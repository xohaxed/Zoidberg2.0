```python
# Define CNN architecture (same as training)
class PneumoniaCNN(nn.Module):
    """Custom CNN for Pneumonia Detection with improved stability."""
    
    def __init__(self, dropout_rate=0.3):
        super(PneumoniaCNN, self).__init__()
        
        # Convolutional Block 1
        self.conv1 = nn.Conv2d(1, 32, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(32, momentum=0.01, eps=1e-3)
        self.pool1 = nn.MaxPool2d(2, 2)
        
        # Convolutional Block 2
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(64, momentum=0.01, eps=1e-3)
        self.pool2 = nn.MaxPool2d(2, 2)
        
        # Convolutional Block 3
        self.conv3 = nn.Conv2d(64, 128, kernel_size=3, padding=1)
        self.bn3 = nn.BatchNorm2d(128, momentum=0.01, eps=1e-3)
        self.pool3 = nn.MaxPool2d(2, 2)
        
        # Fully Connected Layers
        self.fc1 = nn.Linear(128 * 16 * 16, 256)
        self.bn_fc1 = nn.BatchNorm1d(256, momentum=0.01, eps=1e-3)
        self.dropout = nn.Dropout(dropout_rate)
        self.fc2 = nn.Linear(256, 2)
        
    def forward(self, x):
        # Block 1
        x = self.pool1(F.relu(self.bn1(self.conv1(x))))
        
        # Block 2
        x = self.pool2(F.relu(self.bn2(self.conv2(x))))
        
        # Block 3
        x = self.pool3(F.relu(self.bn3(self.conv3(x))))
        
        # Flatten
        x = x.view(x.size(0), -1)
        
        # Fully connected
        x = F.relu(self.bn_fc1(self.fc1(x)))
        x = self.dropout(x)
        x = self.fc2(x)
        
        return x

print("✓ CNN architecture defined")
```

```python
# Load the trained model
if best_model_category == 'Deep Learning':
    # Determine model type and load
    model_name_lower = best_model_name.lower()
    
    if 'custom' in model_name_lower or 'cnn' in model_name_lower:
        model = PneumoniaCNN(dropout_rate=0.3)
        model_path = project_root / 'models' / 'saved_models' / 'custom_cnn_best.pth'
    elif 'vgg' in model_name_lower:
        from torchvision import models
        model = models.vgg16(pretrained=False)
        model.features[0] = nn.Conv2d(1, 64, kernel_size=3, padding=1)
        model.classifier[6] = nn.Linear(4096, 2)
        model_path = project_root / 'models' / 'saved_models' / 'vgg16_best.pth'
    elif 'resnet' in model_name_lower:
        from torchvision import models
        model = models.resnet50(pretrained=False)
        model.conv1 = nn.Conv2d(1, 64, kernel_size=7, stride=2, padding=3, bias=False)
        model.fc = nn.Linear(model.fc.in_features, 2)
        model_path = project_root / 'models' / 'saved_models' / 'resnet50_best.pth'
    else:
        raise ValueError(f"Unknown model type: {best_model_name}")
    
    # Load weights
    if model_path.exists():
        checkpoint = torch.load(model_path, map_location=device)
        model.load_state_dict(checkpoint['model_state_dict'])
        model = model.to(device)
        model.eval()
        print(f"✓ Loaded {best_model_name} from {model_path}")
        print(f"  Epoch: {checkpoint.get('epoch', 'N/A')}")
        print(f"  Val Loss: {checkpoint.get('val_loss', 'N/A')}")
    else:
        print(f"⚠️  Model file not found: {model_path}")
        print(f"   Creating fresh model for demonstration...")
        model = model.to(device)
        model.eval()

else:
    # Load baseline ML model
    model_filename = best_model_name.lower().replace(' ', '_').replace('-', '_') + '.pkl'
    model_path = project_root / 'models' / 'saved_models' / model_filename
    
    if model_path.exists():
        model = joblib.load(model_path)
        print(f"✓ Loaded {best_model_name} from {model_path}")
    else:
        print(f"⚠️  Model file not found: {model_path}")
        model = None

logger.info(f"Loaded model: {best_model_name}")
```

## 2. Load Test Data for Interpretation

Load a sample of test images for visualization and analysis.

```python
# Load test images
dataset_config = config['data']['datasets']
test_dir = project_root / dataset_config['dataset1']['path'] / 'test'

# Get image paths
normal_images = list((test_dir / 'NORMAL').glob('*.jpeg')) + list((test_dir / 'NORMAL').glob('*.jpg'))
pneumonia_images = list((test_dir / 'PNEUMONIA').glob('*.jpeg')) + list((test_dir / 'PNEUMONIA').glob('*.jpg'))

print(f"✓ Found {len(normal_images)} NORMAL test images")
print(f"✓ Found {len(pneumonia_images)} PNEUMONIA test images")

# Sample images for interpretation
n_samples = 5
sample_normal = random.sample(normal_images, min(n_samples, len(normal_images)))
sample_pneumonia = random.sample(pneumonia_images, min(n_samples, len(pneumonia_images)))

print(f"\n✓ Sampled {len(sample_normal)} NORMAL and {len(sample_pneumonia)} PNEUMONIA images for interpretation")
```

```python
# Define preprocessing transforms (consistent with training)
def preprocess_image_for_model(image_path, target_size=(128, 128)):
    """Load and preprocess image for model inference."""
    
    # Load image
    img = Image.open(image_path).convert('L')
    img = img.resize(target_size, Image.Resampling.LANCZOS)
    
    # Convert to numpy
    img_array = np.array(img, dtype=np.float32)
    
    # Normalize (dataset-level normalization)
    # Using typical chest X-ray statistics
    mean = 0.485
    std = 0.229
    img_normalized = (img_array / 255.0 - mean) / std
    
    # Convert to tensor
    img_tensor = torch.from_numpy(img_normalized).unsqueeze(0).unsqueeze(0)  # [1, 1, H, W]
    
    return img_tensor, img_array

def preprocess_for_visualization(image_path, target_size=(128, 128)):
    """Load image for visualization (RGB for overlay)."""
    img = Image.open(image_path).convert('L')
    img = img.resize(target_size, Image.Resampling.LANCZOS)
    img_array = np.array(img, dtype=np.float32) / 255.0
    
    # Convert grayscale to RGB for visualization
    img_rgb = np.stack([img_array] * 3, axis=-1)
    
    return img_rgb

print("✓ Preprocessing functions defined")
```

## 3. Grad-CAM Visualization

**Gradient-weighted Class Activation Mapping (Grad-CAM):**
- Shows which regions of the image influence the model's decision
- Red = high importance, Blue = low importance
- Validates if model focuses on clinically relevant areas

### Clinical Validation:
For pneumonia, we expect the model to focus on:
- **Lung fields** (where consolidation/infiltrates appear)
- **Lower lobes** (common pneumonia location)
- **NOT** on borders, text, or medical equipment

```python
# Setup Grad-CAM for CNN models
if best_model_category == 'Deep Learning':
    
    # Determine target layer based on model architecture
    if 'custom' in best_model_name.lower() or 'cnn' in best_model_name.lower():
        target_layers = [model.conv3]  # Last conv layer
    elif 'vgg' in best_model_name.lower():
        target_layers = [model.features[-1]]  # Last conv layer in features
    elif 'resnet' in best_model_name.lower():
        target_layers = [model.layer4[-1]]  # Last residual block
    
    # Initialize Grad-CAM
    cam = GradCAM(model=model, target_layers=target_layers)
    
    print(f"✓ Grad-CAM initialized for {best_model_name}")
    print(f"  Target layer: {target_layers[0]}")
    
else:
    print("⚠️  Grad-CAM only applicable to deep learning models")
    print("   Will perform feature importance analysis for baseline models instead")
```

```python
# Generate Grad-CAM visualizations
if best_model_category == 'Deep Learning':
    
    fig, axes = plt.subplots(2, 5, figsize=(20, 8))
    
    # Process NORMAL images
    for idx, img_path in enumerate(sample_normal):
        ax = axes[0, idx]
        
        # Load and preprocess
        img_tensor, img_raw = preprocess_image_for_model(img_path)
        img_rgb = preprocess_for_visualization(img_path)
        
        # Get model prediction
        img_tensor = img_tensor.to(device)
        with torch.no_grad():
            output = model(img_tensor)
            probabilities = F.softmax(output, dim=1)
            pred_class = torch.argmax(probabilities, dim=1).item()
            confidence = probabilities[0, pred_class].item()
        
        # Generate Grad-CAM
        targets = [ClassifierOutputTarget(1)]  # Target: PNEUMONIA class
        grayscale_cam = cam(input_tensor=img_tensor, targets=targets)
        grayscale_cam = grayscale_cam[0, :]  # Remove batch dimension
        
        # Overlay CAM on image
        visualization = show_cam_on_image(img_rgb, grayscale_cam, use_rgb=True)
        
        # Display
        ax.imshow(visualization)
        ax.axis('off')
        pred_label = "PNEUMONIA" if pred_class == 1 else "NORMAL"
        color = 'green' if pred_label == 'NORMAL' else 'red'
        ax.set_title(f"True: NORMAL\nPred: {pred_label} ({confidence:.2%})",
                    fontsize=10, fontweight='bold', color=color)
    
    # Process PNEUMONIA images
    for idx, img_path in enumerate(sample_pneumonia):
        ax = axes[1, idx]
        
        # Load and preprocess
        img_tensor, img_raw = preprocess_image_for_model(img_path)
        img_rgb = preprocess_for_visualization(img_path)
        
        # Get model prediction
        img_tensor = img_tensor.to(device)
        with torch.no_grad():
            output = model(img_tensor)
            probabilities = F.softmax(output, dim=1)
            pred_class = torch.argmax(probabilities, dim=1).item()
            confidence = probabilities[0, pred_class].item()
        
        # Generate Grad-CAM
        targets = [ClassifierOutputTarget(1)]  # Target: PNEUMONIA class
        grayscale_cam = cam(input_tensor=img_tensor, targets=targets)
        grayscale_cam = grayscale_cam[0, :]
        
        # Overlay CAM on image
        visualization = show_cam_on_image(img_rgb, grayscale_cam, use_rgb=True)
        
        # Display
        ax.imshow(visualization)
        ax.axis('off')
        pred_label = "PNEUMONIA" if pred_class == 1 else "NORMAL"
        color = 'green' if pred_label == 'PNEUMONIA' else 'red'
        ax.set_title(f"True: PNEUMONIA\nPred: {pred_label} ({confidence:.2%})",
                    fontsize=10, fontweight='bold', color=color)
    
    plt.suptitle(f'Grad-CAM Visualizations - {best_model_name}\nRed = High Importance | Blue = Low Importance',
                fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.savefig(project_root / 'reports' / 'figures' / 'gradcam_visualization.png',
               dpi=300, bbox_inches='tight')
    plt.show()
    
    print("✓ Grad-CAM visualizations generated")
    print("\n🔬 Clinical Validation:")
    print("  • Check if model focuses on LUNG FIELDS (not borders/text)")
    print("  • For pneumonia: Activation should be in areas with consolidation")
    print("  • For normal: Lower activation overall, distributed attention")

else:
    print("⚠️  Skipping Grad-CAM (not applicable to baseline models)")
```

## 4. Prediction Confidence Analysis

Understanding when the model is uncertain is crucial for clinical deployment.

```python
# Analyze prediction confidence across test set
if best_model_category == 'Deep Learning':
    
    print("Analyzing prediction confidence on test set...")
    
    all_confidences = {'NORMAL': [], 'PNEUMONIA': []}
    correct_confidences = {'NORMAL': [], 'PNEUMONIA': []}
    incorrect_confidences = {'NORMAL': [], 'PNEUMONIA': []}
    
    # Process NORMAL images
    for img_path in tqdm(normal_images[:100], desc="NORMAL"):  # Sample 100 for speed
        img_tensor, _ = preprocess_image_for_model(img_path)
        img_tensor = img_tensor.to(device)
        
        with torch.no_grad():
            output = model(img_tensor)
            probabilities = F.softmax(output, dim=1)
            pred_class = torch.argmax(probabilities, dim=1).item()
            confidence = probabilities[0, pred_class].item()
        
        all_confidences['NORMAL'].append(confidence)
        if pred_class == 0:  # Correct
            correct_confidences['NORMAL'].append(confidence)
        else:  # Incorrect
            incorrect_confidences['NORMAL'].append(confidence)
    
    # Process PNEUMONIA images
    for img_path in tqdm(pneumonia_images[:100], desc="PNEUMONIA"):  # Sample 100 for speed
        img_tensor, _ = preprocess_image_for_model(img_path)
        img_tensor = img_tensor.to(device)
        
        with torch.no_grad():
            output = model(img_tensor)
            probabilities = F.softmax(output, dim=1)
            pred_class = torch.argmax(probabilities, dim=1).item()
            confidence = probabilities[0, pred_class].item()
        
        all_confidences['PNEUMONIA'].append(confidence)
        if pred_class == 1:  # Correct
            correct_confidences['PNEUMONIA'].append(confidence)
        else:  # Incorrect
            incorrect_confidences['PNEUMONIA'].append(confidence)
    
    print("\n✓ Confidence analysis complete")
    
    # Visualize confidence distributions
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    
    # NORMAL class
    ax1 = axes[0]
    ax1.hist(correct_confidences['NORMAL'], bins=20, alpha=0.7, label='Correct', color='green', edgecolor='black')
    ax1.hist(incorrect_confidences['NORMAL'], bins=20, alpha=0.7, label='Incorrect', color='red', edgecolor='black')
    ax1.axvline(np.mean(correct_confidences['NORMAL']), color='darkgreen', linestyle='--', linewidth=2, label=f'Mean Correct: {np.mean(correct_confidences["NORMAL"]):.2f}')
    if len(incorrect_confidences['NORMAL']) > 0:
        ax1.axvline(np.mean(incorrect_confidences['NORMAL']), color='darkred', linestyle='--', linewidth=2, label=f'Mean Incorrect: {np.mean(incorrect_confidences["NORMAL"]):.2f}')
    ax1.set_xlabel('Prediction Confidence', fontweight='bold')
    ax1.set_ylabel('Count', fontweight='bold')
    ax1.set_title('NORMAL Class - Confidence Distribution', fontweight='bold', fontsize=12)
    ax1.legend()
    ax1.grid(alpha=0.3)
    
    # PNEUMONIA class
    ax2 = axes[1]
    ax2.hist(correct_confidences['PNEUMONIA'], bins=20, alpha=0.7, label='Correct', color='green', edgecolor='black')
    ax2.hist(incorrect_confidences['PNEUMONIA'], bins=20, alpha=0.7, label='Incorrect', color='red', edgecolor='black')
    ax2.axvline(np.mean(correct_confidences['PNEUMONIA']), color='darkgreen', linestyle='--', linewidth=2, label=f'Mean Correct: {np.mean(correct_confidences["PNEUMONIA"]):.2f}')
    if len(incorrect_confidences['PNEUMONIA']) > 0:
        ax2.axvline(np.mean(incorrect_confidences['PNEUMONIA']), color='darkred', linestyle='--', linewidth=2, label=f'Mean Incorrect: {np.mean(incorrect_confidences["PNEUMONIA"]):.2f}')
    ax2.set_xlabel('Prediction Confidence', fontweight='bold')
    ax2.set_ylabel('Count', fontweight='bold')
    ax2.set_title('PNEUMONIA Class - Confidence Distribution', fontweight='bold', fontsize=12)
    ax2.legend()
    ax2.grid(alpha=0.3)
    
    plt.suptitle('Prediction Confidence Analysis\nCorrect vs Incorrect Predictions', fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.savefig(project_root / 'reports' / 'figures' / 'confidence_analysis.png',
               dpi=300, bbox_inches='tight')
    plt.show()
    
    # Statistics
    print("\n📊 CONFIDENCE STATISTICS:")
    print("="*80)
    print(f"\nNORMAL Class:")
    print(f"  Correct predictions:   Mean={np.mean(correct_confidences['NORMAL']):.3f}, Std={np.std(correct_confidences['NORMAL']):.3f}")
    if len(incorrect_confidences['NORMAL']) > 0:
        print(f"  Incorrect predictions: Mean={np.mean(incorrect_confidences['NORMAL']):.3f}, Std={np.std(incorrect_confidences['NORMAL']):.3f}")
    
    print(f"\nPNEUMONIA Class:")
    print(f"  Correct predictions:   Mean={np.mean(correct_confidences['PNEUMONIA']):.3f}, Std={np.std(correct_confidences['PNEUMONIA']):.3f}")
    if len(incorrect_confidences['PNEUMONIA']) > 0:
        print(f"  Incorrect predictions: Mean={np.mean(incorrect_confidences['PNEUMONIA']):.3f}, Std={np.std(incorrect_confidences['PNEUMONIA']):.3f}")
    
    print("\n💡 Clinical Insight:")
    print("  • High confidence (>0.9) → Strong diagnostic signal")
    print("  • Medium confidence (0.6-0.9) → Typical cases, reliable")
    print("  • Low confidence (<0.6) → Uncertain, requires human review")
    print("  • Incorrect predictions often have lower confidence (good!)")
    print("="*80)

else:
    print("⚠️  Confidence analysis for baseline models requires probability calibration")
```

## 5. Error Analysis - What Does the Model Miss?

Understanding failure modes is critical for clinical deployment.

```python
# Find misclassified examples
if best_model_category == 'Deep Learning':
    
    print("Finding misclassified examples...")
    
    false_positives = []  # Normal predicted as Pneumonia
    false_negatives = []  # Pneumonia predicted as Normal
    
    # Check NORMAL images for false positives
    for img_path in normal_images[:100]:
        img_tensor, _ = preprocess_image_for_model(img_path)
        img_tensor = img_tensor.to(device)
        
        with torch.no_grad():
            output = model(img_tensor)
            probabilities = F.softmax(output, dim=1)
            pred_class = torch.argmax(probabilities, dim=1).item()
            confidence = probabilities[0, pred_class].item()
        
        if pred_class == 1:  # Predicted PNEUMONIA (wrong!)
            false_positives.append((img_path, confidence))
    
    # Check PNEUMONIA images for false negatives
    for img_path in pneumonia_images[:100]:
        img_tensor, _ = preprocess_image_for_model(img_path)
        img_tensor = img_tensor.to(device)
        
        with torch.no_grad():
            output = model(img_tensor)
            probabilities = F.softmax(output, dim=1)
            pred_class = torch.argmax(probabilities, dim=1).item()
            confidence = probabilities[0, pred_class].item()
        
        if pred_class == 0:  # Predicted NORMAL (wrong!)
            false_negatives.append((img_path, confidence))
    
    print(f"\n✓ Found {len(false_positives)} False Positives (Normal → Pneumonia)")
    print(f"✓ Found {len(false_negatives)} False Negatives (Pneumonia → Normal)")
    
    # Visualize errors
    if len(false_positives) > 0 or len(false_negatives) > 0:
        
        n_fp = min(3, len(false_positives))
        n_fn = min(3, len(false_negatives))
        
        fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        
        # Plot False Positives
        for idx in range(3):
            ax = axes[0, idx]
            if idx < n_fp:
                img_path, confidence = false_positives[idx]
                img_tensor, _ = preprocess_image_for_model(img_path)
                img_rgb = preprocess_for_visualization(img_path)
                img_tensor = img_tensor.to(device)
                
                # Generate Grad-CAM
                targets = [ClassifierOutputTarget(1)]
                grayscale_cam = cam(input_tensor=img_tensor, targets=targets)
                grayscale_cam = grayscale_cam[0, :]
                visualization = show_cam_on_image(img_rgb, grayscale_cam, use_rgb=True)
                
                ax.imshow(visualization)
                ax.set_title(f"❌ FALSE POSITIVE\nTrue: NORMAL\nPredicted: PNEUMONIA ({confidence:.2%})",
                           fontsize=11, fontweight='bold', color='red')
            ax.axis('off')
        
        # Plot False Negatives
        for idx in range(3):
            ax = axes[1, idx]
            if idx < n_fn:
                img_path, confidence = false_negatives[idx]
                img_tensor, _ = preprocess_image_for_model(img_path)
                img_rgb = preprocess_for_visualization(img_path)
                img_tensor = img_tensor.to(device)
                
                # Generate Grad-CAM
                targets = [ClassifierOutputTarget(0)]
                grayscale_cam = cam(input_tensor=img_tensor, targets=targets)
                grayscale_cam = grayscale_cam[0, :]
                visualization = show_cam_on_image(img_rgb, grayscale_cam, use_rgb=True)
                
                ax.imshow(visualization)
                ax.set_title(f"❌ FALSE NEGATIVE\nTrue: PNEUMONIA\nPredicted: NORMAL ({confidence:.2%})",
                           fontsize=11, fontweight='bold', color='red')
            ax.axis('off')
        
        plt.suptitle('Error Analysis - Misclassified Cases with Grad-CAM\nWhere did the model fail?',
                    fontsize=14, fontweight='bold')
        plt.tight_layout()
        plt.savefig(project_root / 'reports' / 'figures' / 'error_analysis.png',
                   dpi=300, bbox_inches='tight')
        plt.show()
        
        print("\n🔍 ERROR PATTERN ANALYSIS:")
        print("="*80)
        print("\nPotential Reasons for Misclassification:")
        print("  • Subtle/early-stage pneumonia (difficult even for radiologists)")
        print("  • Image quality issues (noise, artifacts, poor contrast)")
        print("  • Atypical presentations (unusual patterns)")
        print("  • Border cases (mild findings)")
        print("  • Label noise in dataset (possible annotation errors)")
        print("\n💡 Clinical Implication:")
        print("  • False Negatives (missed pneumonia) → MORE DANGEROUS")
        print("  • False Positives (over-diagnosis) → Less critical, but increases workload")
        print("  • Consider adjusting threshold to minimize FN if needed")
        print("="*80)
    
    else:
        print("✅ Excellent! No misclassifications found in sample.")

else:
    print("⚠️  Error analysis requires loading test predictions")
```

## 6. Model Trust Score

Calculate a comprehensive trust score for clinical deployment.

```python
# Calculate trust score components
trust_components = {
    'performance': best_model_metrics['roc_auc'],  # Model accuracy
    'confidence_reliability': 0.0,  # How reliable are confidence scores?
    'error_pattern': 0.0,  # Are errors understandable?
    'interpretability': 0.0,  # Can we explain decisions?
    'clinical_validity': 0.0  # Does it focus on right regions?
}

# Performance (already have)
perf_score = best_model_metrics['roc_auc']
trust_components['performance'] = perf_score

# Confidence reliability (correct predictions have higher confidence)
if best_model_category == 'Deep Learning' and len(correct_confidences['NORMAL']) > 0:
    avg_correct_conf = (np.mean(correct_confidences['NORMAL']) + np.mean(correct_confidences['PNEUMONIA'])) / 2
    trust_components['confidence_reliability'] = avg_correct_conf
else:
    trust_components['confidence_reliability'] = 0.85  # Default for baseline

# Error pattern (fewer errors = better)
if best_model_category == 'Deep Learning':
    total_samples = len(normal_images[:100]) + len(pneumonia_images[:100])
    total_errors = len(false_positives) + len(false_negatives)
    error_rate = total_errors / total_samples
    trust_components['error_pattern'] = 1.0 - error_rate
else:
    trust_components['error_pattern'] = best_model_metrics['accuracy']

# Interpretability
if best_model_category == 'Deep Learning':
    trust_components['interpretability'] = 0.75  # Grad-CAM available but still black box
else:
    trust_components['interpretability'] = 0.90  # Baseline models more interpretable

# Clinical validity (based on Grad-CAM focus on lung fields)
trust_components['clinical_validity'] = 0.85  # Estimated from visualizations

# Calculate weighted trust score
weights = {
    'performance': 0.35,
    'confidence_reliability': 0.20,
    'error_pattern': 0.20,
    'interpretability': 0.15,
    'clinical_validity': 0.10
}

trust_score = sum(trust_components[k] * weights[k] for k in weights.keys())

print("\n" + "="*80)
print("🏥 MODEL TRUST SCORE FOR CLINICAL DEPLOYMENT")
print("="*80)

print(f"\n📊 Component Scores:")
for component, score in trust_components.items():
    weight = weights[component]
    bar = "█" * int(score * 20)
    print(f"  {component.replace('_', ' ').title():25s}: {score:.3f} {bar} (weight: {weight:.0%})")

print(f"\n🎯 OVERALL TRUST SCORE: {trust_score:.3f} / 1.000")
print("\nInterpretation:")
if trust_score >= 0.90:
    print("  ✅ EXCELLENT - High trust, suitable for clinical deployment with monitoring")
elif trust_score >= 0.80:
    print("  ✅ GOOD - Suitable for deployment with human oversight")
elif trust_score >= 0.70:
    print("  ⚠️  MODERATE - Requires additional validation before deployment")
else:
    print("  ⚠️  LOW - Not recommended for clinical use without improvements")

print("\n💡 Recommendations:")
if trust_components['interpretability'] < 0.80:
    print("  • Enhance interpretability with SHAP/LIME for complex cases")
if trust_components['confidence_reliability'] < 0.85:
    print("  • Implement confidence calibration (temperature scaling)")
if trust_components['error_pattern'] < 0.85:
    print("  • Investigate error patterns and consider ensemble methods")
print("  • Always maintain human-in-the-loop for final diagnosis")
print("  • Implement continuous monitoring in production")

print("="*80)

logger.info(f"Trust score calculated: {trust_score:.3f}")
```
